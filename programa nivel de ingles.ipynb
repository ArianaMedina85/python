PROGRAMA PARA SABER EL NIVEL DE INGLES: A1, A2, B1, B2, C1 Y DETRRMINAR SI ES ADMITIDO O RECHAZADO CON CALIFICACION MAYOR A 8.0  - ARIANA IA - ITSLP 08/10/25

```python
#Leer datos
prom = float(input("ingrese promedio: "))
n1 = input("ingrese nivel de ingles (1-5): ")


try:
    nivel_ingles = int(n1)
    if nivel_ingles not in range(1, 6):
        print("Nivel de ingles invalido. Debe ser un numero entre 1 y 5.")
    else:
        if prom < 8.0:
            print("rechazado por bajo promedio")
        else:
            if nivel_ingles > 3:
                print("admitido")
            else:
                print("rechazado por bajo nivel de ingles")

except ValueError:
    print("Entrada invalida para el nivel de ingles. Por favor ingrese un numero.")
```
