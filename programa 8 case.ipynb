Codigo case para operaciones basicas
#FUNDAMENTOS DE PROGRAMACION ariana itslp

```python
n1 = float(input("introduzca el primer numero: "))
n2 = float(input("introduzca el segundo numero: "))
operacion = input("introduzca la operacion a realizar (*,/,+,-,**,): ")

match operacion:
    case "+":
        R = n1 + n2
        print(f"el resultado es: {R:.2f}")
    case "-":
        R = n1 - n2
        print(f"el resultado es: {R:.2f}")
    case "*":
        R = n1 * n2
        print(f"el resultado es: {R:.2f}")
    case "/":
        R = n1 / n2
        print(f"el resultado es: {R:.2f}")
    case "**":
        R = n1 ** n2
        print(f"el resultado es: {R:.2f}")

    case _:
        print("\n Operación no válida por favor digite una operacion correcta ")
```
