```python
import numpy as np
import pandas as pd # Importamos pandas

datos = np.zeros((100, 5))
for i in range(1, 101):
    R = np.sqrt(i) # Raíz cuadrada (sqr)
    S = np.sin(i)  # Seno (sen)
    C = np.cos(i)  # Coseno (cos)
    T = np.tan(i)  # Tangente (tan)
    datos[i-1] = [i, R, S, C, T] # Se agregó 'i' para que coincida con las 5 columnas

# Creamos un DataFrame de pandas para mostrar los datos como una tabla
df = pd.DataFrame(datos, columns=['I', 'Sqr(I)', 'Sen(I)', 'Cos(I)', 'Tan(I)'])
# Formateamos todas las columnas numéricas a dos decimales para la visualización
display(df.round(2))

```
