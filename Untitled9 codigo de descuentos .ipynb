#CODIGO QUE CALCULE EL PRECIO TOTAL Y AHORRO TOTAL AL COMPRAR TRES PRODUCTOS CON DESCUENTOS, LACTEOS CON 30%, ELECTRONICOS CON 50% Y PAPELERIA CON 80% DE DESCUENTO


```python
#Declaramos los valores de descuento
descuento_lacteos = 0.3
descuento_electronicos= 0.5
descuento_papeleria = 0.8

#declaramos las variables de precios
Pl = float(input("Ingrese el precio de los productos lacteos: $"))
Pe = float(input("Ingrese el precio de los productos electronicos: $"))
Pp = float(input("Ingrese el precio de los productos de papeleria: $"))

#realizamos operaciones debidas
ahorro_lacteos = Pl * descuento_lacteos
precio_final_lacteos = Pl - ahorro_lacteos

ahorro_electronicos = Pe * descuento_electronicos
precio_final_electronicos = Pe - ahorro_electronicos

ahorro_papeleria = Pp * descuento_papeleria
precio_final_papeleria = Pp - ahorro_papeleria

# Mostramos los resultados por categoria
print(f"\n--- Resultados por Categoria ---")
print(f"Lacteos:")
print(f"  Precio original: ${Pl:.2f}")
print(f"  Ahorro: ${ahorro_lacteos:.2f}")
print(f"  Precio final: ${precio_final_lacteos:.2f}")

print(f"\nElectronicos:")
print(f"  Precio original: ${Pe:.2f}")
print(f"  Ahorro: ${ahorro_electronicos:.2f}")
print(f"  Precio final: ${precio_final_electronicos:.2f}")

print(f"\nPapeleria:")
print(f"  Precio original: ${Pp:.2f}")
print(f"  Ahorro: ${ahorro_papeleria:.2f}")
print(f"  Precio final: ${precio_final_papeleria:.2f}")
```
