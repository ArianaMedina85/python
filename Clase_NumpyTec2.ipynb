### Seguimos con índices de numpy: Booleanos como índices ###

En algunos casos resulta útil seleccionar elementos en función de que cumplan algún condicional. Supongamos por ejemplo que tenemos un arreglo que contiene los tiempos registrados por las corredoras de 100 mts del equipo del IPICYT. Como no podemos guardar arreglos más que con un tipo de información (por las características de numpy), vamos a generar dos arreglos, uno con nombres y otro con los 5 últimos tiempos registrados. Cada uno de los renglones corresponde al registro de los nombres, considerando el mismo orden en que aparecen en el arreglo.

```python
nombres = np.array(["Lisa", "Margaret", "Marjorie", "Patricia", "Marjorie"])
tiempos = np.array([[11.09, 11.2, 11.12, 11.23, 11.21],
                   [11.12, 11.17, 11.16, 11.13, 11.18],
                   [11.04, 11.10, 11.06, 11.08, 11.04],
                   [11.25, 11.27, 11.27, 11.29, 11.21],
                   [11.29, 11.32, 11.3, 11.31, 11.31]])
```

Puedo utilizar condicionales para generar un arreglo de booleanos

```python
print(nombres == "Marjorie")
print()
print(tiempos < 11.10)
```

pero aún más interesante es que puedo utilizar el arreglo de booleanos para seleccionar elementos de otro arreglo (funcionan como una especie de apagador). __OJO__ las dimensiones del arreglo índice y del arreglo del cual selecciono elementos, deben coincidir.

```python
print(nombres == "Marjorie")
print()
print(tiempos)
print()
print(tiempos[nombres == "Marjorie"]) #escoge los rangos o intervalos se escogen
```

```python
print(tiempos)
print()
print(nombres != "Marjorie")
print()
print(tiempos[nombres != "Marjorie"])
```

```python
print(tiempos)
print()
print(tiempos  < 11.10)
print()
print(tiempos[tiempos < 11.10])
```

### Seguimos con índices de numpy: Fancy indexing ###

En numpy se pueden utilizar arreglos de enteros como índices, y a esta idea es a la que se le llama _Fancy indexing_.

__Ejemplo__ , Utilicemos una lista y un ndarray para seleccionar elementos de un arreglo de numpy (ndarray).

```python
# Primero hacemos una lista de los primeros 5 números naturales impares

lista = [i for i in range(1,10,2)]
print(lista)
```

```python
# Luego hacemos un ndarray de los primeros 5 números naturales pares (cast)

ndarr = np.array([i for i in range(0,10,2)])
print(ndarr)
```

```python
# Finalmente, hacemos un arreglo de numpy de 10x5 donde cada renglón corresponda a un número entero distinto
# Observar el uso anidado de list comprehension

arreglonp = np.array([[i + j*0.1 for j in range(5)] for i in range(10)])
print(arreglonp, type(arreglonp))
```

Utilicemos la selección para obtener renglones con índices pares e impares

```python
# Primero impares
impares = arreglonp[[1, 3, 5, 7, 9]]
print(impares)
```

```python
# Después pares
ndarr = np.array([i for i in range(0,10,2)])
print(ndarr, type(ndarr))

#pares = arreglonp[ndarr]
pares = arreglonp[[0, 2, 4, 6, 8]]
#pares = arreglonp[(0, 0)] # Ojo, una tupla como argumento representa otra cosa
print(pares)
```

También podemos utilizar combinaciones de enteros para seleccionar, como parejas de enteros en el caso de tener un arreglo de dos dimensiones.

```python
arreglonp = np.array([[i + j*0.1 for j in range(5)] for i in range(10)])
#print(arreglonp, type(arreglonp))

print(arreglonp)
print()

print("Ahora los elementos de una diagonal:\n")

diagonal = arreglonp[[0,1,2,3],[1,2,3,4]]
print(diagonal)
print()
```

```python
# OJO
diagonal2 = arreglonp[[0,0]]
print(diagonal2)
print()

# Así que:
diagonal2 = arreglonp[[0,1,2],[0,1,2]]
print(diagonal2)
print()
```

### Funciones universales ###

Las _ufunc_ o _funciones universales_ son funciones que operan sobre arreglos de numpy, y que operan bajo la lógica de elemento-a-elemento.

Existen funciones ufunc unarias binarias, es decir, que operan sobre uno o dos argumentos, respectivamente.

__Ejemplos__ con operadores unarios

```python
# Creamos un arreglo
ar1 = np.arange(1,50,2)
print(ar1)
print("Arreglo al cuadrado")
# Elevo al cuadrado
ar2 = ar1*ar1
print(ar2)
# Raiz cuadrada de los elementos
print("\nUtilizando la ufunc sqrt ... ")
print(np.sqrt(ar2))
```

```python
# Creamos un arreglo
ar1 = np.arange(1,20,2)
print("Arreglo inicial")
print(ar1)
# Logaritmo natural de cada uno de los elementos
print("\nUtilizando la ufunc log ... ")
print(np.log(ar1))

#import math
#print(math.log(ar1))
```

__Ejemplos__ , ahora con operadores binarios

```python
# Creamos dos arreglos
ar1 = np.random.rand(1,5)
ar2 = np.random.rand(5)
print("Arreglo 1")
print(ar1)
print("\nArreglo 2")
print(ar2)

# Obtengo el máximo entre ambos arreglos
print("\nUtilizando la ufunc maximum ... ")
print(np.maximum(ar1, ar2))
```

### Estadística básica trabajando arreglos ###

Numpy incluye una gran cantidad de funciones matemáticas estadísticas que pueden operar
sobre todos los elementos de un arreglo, o sobre los datos a lo largo de un eje en particular. Veamos un ejemplo.

```python
# Genero datos aleatorios con numpy.random
data = np.random.randn(6,5)
print(data)
```

Podemos obtener el promedio sobre todo el conjunto de datos

```python
print(np.mean(data))
```

O podemos obtener el promedio sobre uno de los ejes en particular

```python
data = np.random.randn(6,5)
print(data)
print()
# obtengo promedio sobre el primer eje
print("Esto resulta de promediar sobre el primer eje")
print(np.mean(data, axis=0).shape)
print(np.mean(data, axis=0))
print()
# obtengo promedio sobre el segundo eje
print("Esto resulta de promediar sobre el segundo eje")
print(np.mean(data, axis=1).shape)
print(np.mean(data, axis=1))
```
