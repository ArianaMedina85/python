## Numpy ##

Numpy es una biblioteca que provee herramientas muy útiles para cómputo científico. Entre otras cosas, contiene:

* extensión para operaciones de alto desempeño con arreglos multidimensionales

#### Características de los arreglos multidimensionales ####

Algunos atributos importantes de los arreglos de numpy (ndarray):
    
    * ndarray.ndim - Número de dimensiones (número de ejes) del arreglo.
    * ndarray.shape - Tamaño de cada dimensión del arreglo, expresados como una tupla de enteros. Cada entero expresa el tamaño de una dimensión.
    * ndarray.size - Número total de elemento del arreglo. Es el producto del tamaño de cada dimensión.  
    * ndarray.dtype - Objeto que describe el tipo de elementos en el arreglo. Es posible especificarla al crear un arreglo. Además de los tipos regulares de Python, numpy incluye tipos de elementos como numpy.int32, numpy.int16, y numpy.float64, entre otros.
    * ndarray.itemsize - Tamaño en bytes de cada elemento del arreglo

### Indizado y slicing ###

Estos arreglos pueden ser n-dimensionales. Recordemos que las dimensiones de un arreglo se pueden consultar mediante arreglo.ndim

Cada dimensión tiene un tamaño, que corresponde al número de elementos que componen a esa dimensión. Por ejemplo, np.ones((3,4,2)) tiene 3 dimensiones, donde la primera dimensión tienen una longitud de 3, la segunda tiene una longitud de 4 y la tercera tienen una longitud de 2.  

__Ejemplo__ Para un arreglo uni-dimensional, el indizado y slicing funcionan de la misma forma que con una lista.

```python
# Arreglo de 3x1
arreglo = np.array([1.5, 2.5, 3.5, 2.1, 3.1, 4.1, 0.7, 1.7, 2.7])

print("Estos son todos los elementos del arreglo:")
print(arreglo)
print("")
print("Este es el 3er elemento del arreglo:")
print(arreglo[2])
print("")
print("Ahora van los primeros cuatro elementos:")
print(arreglo[0:4])
print("")
print("Ahora van los últimos cuatro elementos:")
print(arreglo[-4:])
```

__Ejemplo 2__ Generemos 100 puntos aleatorios dentro de un cubo cuyos lados valen 1, una de sus aristas coincide con el origen y se encuentra dentro del cuadrante definido por x > 0, y > 0, z > 0 en un espacio 3D. Calculemos después la distancia promedio de estos puntos con respecto al punto (1,0,0), y comparemos el tiempo que se necesita para calcular las distancias con numpy contra una forma más "tradicional".

```python
# observar funciones de numpy random.random y linalg.norm
import numpy as np
from random import random as rconv


def funnp():
    pos = np.random.random((100,3)) # <- 100 elementos con 3 componentes "aleatorias"
    vec = np.array([1,0,0])
    pos -= vec
    lisdis = np.linalg.norm(pos, axis=1) # a
    return lisdis

def funtrad():
    listav = []
    for i in range(100):
        vec = []
        for j in range(3):
            vec.append(rconv())
        listav.append(vec)
    for vector in listav:
        vector[0] -= 1.0
        vector[1] -= 0.0
        vector[2] -= 0.0
    lisdis = []
    for vector in listav:
        dist = 0.0
        for i in range(3):
            dist += vector[i]*vector[i]
        lisdis.append(np.sqrt(dist))
    return lisdis


%timeit funnp()
%timeit funtrad()
```
