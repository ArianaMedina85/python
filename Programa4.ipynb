Programa para determinar si un alumno aprueba o reprueba - Ariana **ITSLP** *FUNDAMENTOS DE PROGRAMACION*


```python
#Leer datos
num1 = float(input("Ingrese la primera nota: "))
num2 = float(input("Ingrese la segunda nota: "))
num3 = float(input("Ingrese la tercera nota: "))
#Mostramos operacion
promedio = (num1 + num2 + num3) / 3
#Condicionamos el resultado
if promedio >= 7:
  print(f"El alumno aprueba con: {promedio:.2f}")
else:
  print(f"El alumno reprueba con: {promedio:.2f}")
```
