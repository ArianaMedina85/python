## 1. Ciclo `for` – Recorrer secuencias
### _Ejemplo 1: Suma de elementos de una lista_
# Lista de números

```python
numeros = [1, 2, 3, 4, 5]
suma = 0

for num in numeros:
  suma += num

print(f"Suma total: {suma}") # Salida: Suma total: 15
```

### Ejemplo 2: Como crear una lista de cuadrados

```python
cuadrados = []
for i in range(1, 6): #para cada elemento i en el rango de 1 al 6
  cuadrados.append(i ** 2)
print(f"Cuadrados: {cuadrados}") # Salida: Cuadrados: [1, 4, 9, 16, 25]
```

Modifica el código anterior para obtener el número y su respectivo cuadrado

```python
cuadrados = []
for i in range(1, 6):
  cuadrados.append(i ** 2)
  print(f"El número {i} y su cuadrado es: {i ** 2}")
print(f"Cuadrados: {cuadrados}")
```

### Ejemplo 3: Filtrar números pares (útil en preprocesamiento de datos)

```python

numeros = [10, 21, 22, 33, 40]
pares = []

for n in numeros: # 0 n=10, 1 n=21, 2 n=22
  if n % 2 == 0:
    pares.append(n)

print(f"Números pares: {pares}") # Salida: Números pares: [10, 22, 40]
```

## 2. Ciclo `while` – Repetición basada en condición

### Ejemplo 1: Contar hasta un número dado

```python
contador = 0 #incrementar
limite = 5 #tope

while contador < limite:
  print(f"Contador: {contador}")
  contador += 1 #contador = contador + 1, contador = 2
```

### Ejemplo 2: Adivinar un número (juego simple)

```python
import random
numero_secreto = random.randint(1, 10)
intento = None

while intento != numero_secreto:
  intento = int(input("Adivina el número (entre 1 y 10): "))
  if intento < numero_secreto:
    print("Más alto!")
  elif intento > numero_secreto:
    print("Más bajo!")

print("¡Correcto! El número era", numero_secreto)
```

### Ejercicio 1: Calcular promedio de una lista
Crea una función `calcular_promedio(lista)` que use un ciclo `for` para sumar los elementos y luego divida entre la cantidad de elementos.

```python
#Escribe aqui tu codigo
lista = [10, 20, 30, 40, 50]
suma = 0

for elemento in lista:
  suma += elemento

promedio = suma / len(lista)
print(f"El promedio de {lista} es: {promedio}")
```

### Ejercicio 2: Encontrar el máximo valor en una lista sin usar `max()`
Usa un ciclo `for` para iterar y guardar el valor máximo encontrado.

```python
#Escribe aqui tu codigo
lista = [3, 1, 4, 1, 5, 9, 2, 6]

# Asumimos que el primer elemento es el máximo inicialmente
max_valor = lista[0]

# Repetimos sobre el resto de la lista para comparar y encontrar el verdadero máximo
for numero in lista:
  if numero > max_valor:
    max_valor = numero

print(f"El valor máximo en la lista {lista} es: {max_valor}")
```


### Ejercicio 3: Generar secuencia de Fibonacci hasta N
Usa un ciclo `while` para generar la secuencia de Fibonacci hasta un número `N` dado.

```python
n = int(input("Ingresa el número hasta el cual quieres la secuencia de Fibonacci: ")) #7

a, b = 0, 1
secuencia_fibonacci = [] #lista [] vacia

while a <= n: #mientras que ..
  secuencia_fibonacci.append(a) #se va a a;adir un elemento a la lista secuencia_fibonaci con el metodo append
  #[0]
  a, b = b, a + b

print(f"Secuencia de Fibonacci hasta {n}: {secuencia_fibonacci}")
```

### Ejercicio 4: Contar palabras en una frase
Pedir al usuario una frase y contar cuántas veces aparece cada palabra (útil para procesamiento de lenguaje natural).


```python
frase = input("Por favor, introduce una frase: ")

# Convertir la frase a minúsculas
frase = frase.lower()

# Dividir la frase en palabras usando espacios como delimitador
palabras = frase.split()

# Contar la frecuencia de cada palabra
conteo_palabras = {}
for palabra in palabras:
  # Eliminamos puntuación básica de los extremos de la palabra para simplificar
  palabra_limpia = palabra.strip ('.,!?;:"()')
  if palabra_limpia:
    conteo_palabras[palabra_limpia] = conteo_palabras.get(palabra_limpia, 0) + 1

print("Conteo de palabras:")
for palabra, cuenta in conteo_palabras.items():
  print(f"'{palabra}': {cuenta}")
```

```python
#aqui va tu codigo
frase=input("Por favor, introduce una frase: ")
frase = frase.lower()
palabras = frase.split()
conteo_palabras = {}

for palabra in palabras:
  if palabra in conteo_palabras:
    conteo_palabras[palabra] += 1
  else:
    conteo_palabras[palabra] = 1
print('conteo de palabras:')
for palabra, veces in conteo_palabras.items():
  print(f'{palabra}:{veces}')
```
