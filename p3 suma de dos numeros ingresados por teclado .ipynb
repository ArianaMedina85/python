```python
num1_str = input("Enter the first number: ")
num2_str = input("Enter the second number: ")

# Convert the input strings to integers
num1 = int(num1_str)
num2 = int(num2_str)

sum_result = num1 + num2
print("The sum of", num1, "and", num2, "is:", sum_result)
```
