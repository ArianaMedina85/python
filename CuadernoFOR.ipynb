## 1. Ciclo `for` – Recorrer secuencias
### _Ejemplo 1: Suma de elementos de una lista_
# Lista de números

```python
numeros = [1, 2, 3, 4, 5]
suma = 0

for num in numeros:
  suma += num

print(f"Suma total: {suma}") # Salida: Suma total: 15
```

### Ejemplo 2: Como crear una lista de cuadrados

```python
cuadrados = []
for i in range(1, 6):
  cuadrados.append(i ** 2)
print(f"Cuadrados: {cuadrados}") # Salida: Cuadrados: [1, 4, 9, 16, 25]
```

Modifica el código anterior para obtener el número y su respectivo cuadrado

```python
cuadrados = []
for i in range(1, 6):
  cuadrados.append(i ** 2)
  print(f"El número {i} y su cuadrado es: {i ** 2}")
print(f"Cuadrados: {cuadrados}")
```

### Ejemplo 3: Filtrar números pares (útil en preprocesamiento de datos)

```python

numeros = [10, 21, 22, 33, 40]
pares = []

for n in numeros:
  if n % 2 == 0:
    pares.append(n)

print(f"Números pares: {pares}") # Salida: Números pares: [10, 22, 40]
```

## 2. Ciclo `while` – Repetición basada en condición

### Ejemplo 1: Contar hasta un número dado

```python
contador = 0
limite = 5

while contador < limite:
  print(f"Contador: {contador}")
  contador += 1
```

### Ejemplo 2: Adivinar un número (juego simple)

```python
import random
numero_secreto = random.randint(1, 10)
intento = None

while intento != numero_secreto:
  intento = int(input("Adivina el número (entre 1 y 10): "))
  if intento < numero_secreto:
    print("Más alto!")
  elif intento > numero_secreto:
    print("Más bajo!")

print("¡Correcto! El número era", numero_secreto)
```
