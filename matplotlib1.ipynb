```python
# @title Instalación y Importación
!pip install matplotlib  # Solo si necesitas instalar (ya viene con Colab)
import matplotlib.pyplot as plt
import numpy as np

```

```python
# @title 1. Gráfico Simple (Línea)
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sen(x)', color='blue', linestyle='-')
plt.title('Gráfico de sen(x)')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend()
plt.grid(True)
plt.show()

```

```python
# @title 2. Múltiples Gráficos en una Figura (Subplots)
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].plot(x, np.sin(x), 'r-')
axs[0, 0].set_title('sen(x)')

axs[0, 1].plot(x, np.cos(x), 'g--')
axs[0, 1].set_title('cos(x)')

axs[1, 0].scatter(x, np.random.rand(len(x)), color='orange')
axs[1, 0].set_title('Puntos Aleatorios')

axs[1, 1].bar(['A', 'B', 'C'], [3, 7, 5], color='purple')
axs[1, 1].set_title('Barras')

plt.tight_layout()
plt.show()

```

```python
# @title 3. Gráfico de Dispersión (Scatter)
x = np.random.randn(100)
y = np.random.randn(100)

plt.scatter(x, y, c=x, cmap='viridis', alpha=0.7)
plt.colorbar(label='Color')
plt.title('Gráfico de Dispersión')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()

```

```python
# @title 4. Gráfico de Barras
categorias = ['A', 'B', 'C', 'D']
valores = [4, 7, 2, 5]

plt.bar(categorias, valores, color=['red', 'green', 'blue', 'orange'])
plt.title('Gráfico de Barras')
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.show()

```

```python
# @title 5. Histograma
datos = np.random.normal(0, 1, 1000)

plt.hist(datos, bins=30, edgecolor='black', alpha=0.7, color='skyblue')
plt.title('Histograma')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.show()

```

```python
# @title 6. Personalización Avanzada
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sen(x)', linewidth=2, color='tab:blue')
plt.plot(x, y2, label='cos(x)', linewidth=2, color='tab:orange', linestyle='--')

plt.title('Gráficos Personalizados', fontsize=16)
plt.xlabel('Eje X', fontsize=12)
plt.ylabel('Eje Y', fontsize=12)
plt.legend(loc='best', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```
