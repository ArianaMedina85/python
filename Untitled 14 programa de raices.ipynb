Programa que me calcula el tipo de raices por discriminante con la formula : Δ=b²-4ac

```python
# Solicita al usuario que ingrese los valores de a, b y c
a = float(input("Ingresa el valor de a: "))
b = float(input("Ingresa el valor de b: "))
c = float(input("Ingresa el valor de c: "))

# Calcula el discriminante
Δ = b**2 - 4*a*c

# Imprime el resultado
print(f"El discriminante (Δ) es: {Δ}")

# Determina el tipo de raíces
if Δ > 0:
    print("Las raíces son reales y distintas.")
elif Δ == 0:
    print("Las raíces son reales e iguales.")
else:
    print("Las raíces son complejas y conjugadas.")
```
